```python
import numpy as np
import matplotlib.pyplot as plt
from opfunu.cec_based import F192014

# Define the number of dimensions for the problem
ndim = 100
f0 = F192014 (ndim=ndim)

# Objective function
def objective_function(x):
    return f0.evaluate(x)

# Initialize a population of vectors (each solution is a vector of ndim dimensions)
def initialize_population(pop_size, ndim, lower_bound, upper_bound):
    return np.random.uniform(lower_bound, upper_bound, (pop_size, ndim))

# Evaluate the fitness of each neuron (lower is better)
def evaluate_fitness(population):
    return np.array([objective_function(x) for x in population])

# Update synaptic weights based on fitness and distance
def update_synaptic_weights(population, fitness, synaptic_weights, learning_rate):
    for i in range(len(population)):
        for j in range(len(population)):
            if i != j:
                distance = np.linalg.norm(population[i] - population[j])  # Use vector distance (Euclidean)
                weight_change = learning_rate * (fitness[i] - fitness[j]) / (distance + 1e-9)
                synaptic_weights[i][j] += weight_change
                synaptic_weights[j][i] += weight_change  # Symmetric update

# Move neurons based on synaptic influence
def move_neurons(population, synaptic_weights, global_best, local_best):
    new_population = np.copy(population)
    for i in range(len(population)):
        influence = np.sum(synaptic_weights[i][:, np.newaxis] * (population - population[i]), axis=0)
        # Include guidance by global and local best solutions
        move = (
            influence +
            0.1 * (global_best - population[i]) +  # Global best guidance
            0.05 * (local_best - population[i])    # Local best guidance
        )
        new_population[i] += move + np.random.uniform(-0.01, 0.01, population.shape[1])  # Small random perturbation
    return new_population

# Adaptive pruning of weak synapses
def prune_synapses(synaptic_weights, prune_threshold, iteration, max_iter):
    adaptive_threshold = prune_threshold * (1 - iteration / max_iter)  # Decay threshold over time
    synaptic_weights[np.abs(synaptic_weights) < adaptive_threshold] = 0

# Neuron Synapse Optimization (NSO) Algorithm
def improved_nso(pop_size, ndim, lower_bound, upper_bound, max_iter, learning_rate, prune_threshold):
    # Initialize population and synaptic weights
    population = initialize_population(pop_size, ndim, lower_bound, upper_bound)
    synaptic_weights = np.zeros((pop_size, pop_size))

    # Track global and local best solutions
    fitness = evaluate_fitness(population)
    global_best_idx = np.argmin(fitness)
    global_best_fitness = fitness[global_best_idx]
    global_best_solution = population[global_best_idx]

    # Initialize local_best_solution to the initial global best solution
    local_best_fitness = global_best_fitness
    local_best_solution = global_best_solution

    convergence_curve = []  # Track global best fitness over iterations


    for iteration in range(max_iter):
        # Update synaptic weights
        update_synaptic_weights(population, fitness, synaptic_weights, learning_rate)

        # Move neurons
        population = move_neurons(population, synaptic_weights, global_best_solution, local_best_solution)

        # Prune weak synapses
        prune_synapses(synaptic_weights, prune_threshold, iteration, max_iter)

        # Evaluate fitness of updated population
        fitness = evaluate_fitness(population)

        # Update global and local best solutions
        current_local_best_idx = np.argmin(fitness)
        current_local_best_fitness = fitness[current_local_best_idx]
        current_local_best_solution = population[current_local_best_idx]

        if current_local_best_fitness < global_best_fitness:
            global_best_fitness = current_local_best_fitness
            global_best_solution = current_local_best_solution

        # Update local best for the current iteration
        local_best_fitness = current_local_best_fitness
        local_best_solution = current_local_best_solution

        # Record convergence
        convergence_curve.append(global_best_fitness)

        # Print progress every 100 iterations
        if iteration % 100 == 0 or iteration == max_iter - 1:
            print(f"Iteration {iteration}: Global Best Fitness = {global_best_fitness}")

    return global_best_fitness, convergence_curve

# Parameters
pop_size = 60
lower_bound = -100.0
upper_bound = 100.0
max_iter = 1000
learning_rate = 0.1  # Increased initial learning rate for dynamic adjustment
prune_threshold = 0.02  # Initial threshold for pruning synapses

# Run Improved NSO Algorithm
final_fitness, convergence_curve = improved_nso(
    pop_size, ndim, lower_bound, upper_bound, max_iter, learning_rate, prune_threshold
)

# Plot Convergence Curve
plt.figure(figsize=(10, 6))
plt.plot(convergence_curve, label="Improved NSO Global Best", linestyle='--', color='green')
plt.title("Convergence Curve - Improved NSO(F212014)")
plt.xlabel("Iteration")
plt.ylabel("Fitness")
plt.legend()
plt.grid(True)
plt.show()

# Print Final Results
print(f"Final Global Best Fitness (Improved NSO): {final_fitness}")
```

```python
import gymnasium as gym
import numpy as np
from stable_baselines3 import PPO
from stable_baselines3.common.vec_env import DummyVecEnv, VecNormalize
from opfunu.cec_based import F192014
class OptimizationEnv(gym.Env):
    def __init__(self, ndim=10):
        super(OptimizationEnv, self).__init__()
        self.ndim = ndim
        self.cec_function = F192014 (ndim=self.ndim)

        # Define proper Box spaces
        self.action_space = gym.spaces.Box(
            low=-100,
            high=100,
            shape=(self.ndim,),
            dtype=np.float32
        )
        self.observation_space = gym.spaces.Box(
            low=-100,
            high=100,
            shape=(self.ndim,),
            dtype=np.float32
        )

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        # Generate random initial state
        self.state = np.random.uniform(-100, 100, size=(self.ndim,))
        return self.state, {}

    def step(self, action):
        # Ensure action is 1D array of length ndim
        action = np.array(action).reshape(-1)

        # Calculate fitness using CEC function
        fitness = self.cec_function.evaluate(action).item()

        # Convert to reward (PPO maximizes, so negate)
        reward = -fitness

        # For continuous tasks, never done
        done = False

        return action, reward, done, False, {}

# Environment setup
ndim = 10
env = OptimizationEnv(ndim)
vec_env = DummyVecEnv([lambda: env])
vec_env = VecNormalize(vec_env, norm_obs=True, norm_reward=True)

# Model configuration
model = PPO(
    "MlpPolicy",
    vec_env,
    verbose=1,
    learning_rate=3e-4,
    n_steps=2048,
    batch_size=64,
    n_epochs=10,
    gamma=0.99,
    gae_lambda=0.95,
    ent_coef=0.0,
    clip_range=0.2
)

# Track global best
global_best_action = None
global_best_fitness = np.inf

# Training phase
print("Starting training...")
model.learn(total_timesteps=50000)

# Evaluation phase
print("\nEvaluating trained model:")
obs, _ = env.reset()
for ep in range(5):
    total_reward = 0
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, done, _, _ = env.step(action)
    total_reward += reward
    fitness_score = -total_reward

    # Track global best solution
    if fitness_score < global_best_fitness:
        global_best_fitness = fitness_score
        global_best_action = action

    print(f"Episode {ep+1}")
    print(f"Best solution: {action}")
    print(f"Fitness score: {fitness_score}\n")

# Report global best solution and fitness
print("Global best solution found:")
print(f"Best action: {global_best_action}")
print(f"Best fitness: {global_best_fitness}")
```

RL (Deep Q- Learning)

```python
import gym
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from collections import deque
import random
from opfunu.cec_based import F192014

# Define the custom environment
class OptimizationEnv(gym.Env):
    def __init__(self, ndim=10):
        super(OptimizationEnv, self).__init__()
        self.ndim = ndim
        self.cec_function = F192014(ndim=self.ndim)

        self.action_space = gym.spaces.Box(low=-100, high=100, shape=(self.ndim,), dtype=np.float32)
        self.observation_space = gym.spaces.Box(low=-100, high=100, shape=(self.ndim,), dtype=np.float32)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        self.state = np.random.uniform(-100, 100, size=(self.ndim,))
        return self.state.astype(np.float32), {}

    def step(self, action):
        action = np.clip(action, -100, 100).reshape(-1).astype(np.float32)
        fitness = self.cec_function.evaluate(action).item()
        reward = -fitness
        done = False
        return action, reward, done, False, {}

# Define the Q-network
class QNetwork(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(QNetwork, self).__init__()
        self.fc = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 128),
            nn.ReLU(),
            nn.Linear(128, output_dim)
        )

    def forward(self, x):
        return self.fc(x)

# DQN Agent
class DQNAgent:
    def __init__(self, state_dim, action_dim):
        self.state_dim = state_dim
        self.action_dim = action_dim
        self.q_net = QNetwork(state_dim, action_dim)
        self.target_net = QNetwork(state_dim, action_dim)
        self.optimizer = optim.Adam(self.q_net.parameters(), lr=1e-3)
        self.criterion = nn.MSELoss()
        self.memory = deque(maxlen=10000)
        self.batch_size = 64
        self.gamma = 0.99
        self.epsilon = 1.0
        self.epsilon_decay = 0.995
        self.epsilon_min = 0.05
        self.update_target_steps = 100
        self.steps_done = 0

    def get_action(self, state):
        if np.random.rand() < self.epsilon:
            return np.random.uniform(-100, 100, size=(self.action_dim,))
        state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0)
        with torch.no_grad():
            q_values = self.q_net(state_tensor)
        return q_values.squeeze().numpy()

    def store_transition(self, transition):
        self.memory.append(transition)

    def train_step(self):
        if len(self.memory) < self.batch_size:
            return
        batch = random.sample(self.memory, self.batch_size)
        states, actions, rewards, next_states = zip(*batch)

        states = torch.tensor(np.array(states), dtype=torch.float32)
        actions = torch.tensor(np.array(actions), dtype=torch.float32)
        rewards = torch.tensor(np.array(rewards), dtype=torch.float32).unsqueeze(1)
        next_states = torch.tensor(np.array(next_states), dtype=torch.float32)

        q_values = self.q_net(states)
        next_q_values = self.target_net(next_states).detach()

        # For continuous actions, approximate best next action's Q-value using max Q
        max_next_q = next_q_values.max(dim=1, keepdim=True).values
        target = rewards + self.gamma * max_next_q

        loss = self.criterion(q_values, target)

        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

        self.steps_done += 1
        if self.steps_done % self.update_target_steps == 0:
            self.target_net.load_state_dict(self.q_net.state_dict())

        if self.epsilon > self.epsilon_min:
            self.epsilon *= self.epsilon_decay

# Main training loop
ndim = 10
env = OptimizationEnv(ndim)
agent = DQNAgent(state_dim=ndim, action_dim=ndim)

global_best_fitness = float('inf')
global_best_action = None

for episode in range(200):
    state, _ = env.reset()
    total_reward = 0

    for step in range(1):
        action = agent.get_action(state)
        next_state, reward, done, _, _ = env.step(action)
        agent.store_transition((state, action, reward, next_state))
        agent.train_step()
        state = next_state
        total_reward += reward

    fitness = -total_reward
    if fitness < global_best_fitness:
        global_best_fitness = fitness
        global_best_action = action

    if episode % 10 == 0:
        print(f"Episode {episode}, Fitness: {fitness:.4f}, Epsilon: {agent.epsilon:.3f}")

print("\nBest solution found:")
print("Best action:", global_best_action)
print("Fitness:", global_best_fitness)
```
